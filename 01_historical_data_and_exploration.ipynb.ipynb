# AMD and NVIDIA: Statistical Analysis & Pairs Trading

## Preliminary Experiment: Returns and Volatility

Before retrieving historical market data, I explored how stock returns
and volatility are calculated using hypothetical price observations.

```python
import statistics

amd_prices = [150.25, 160, 145, 165, 200]
nvidia_prices = [180.50, 195, 190, 210, 220]

amd_returns = []
nvidia_returns = []

for day in range(len(amd_prices)-1):
    curr_return = (amd_prices[day+1] - amd_prices[day]) / amd_prices[day]
    amd_returns.append(curr_return)

for day in range(len(nvidia_prices)-1):
    curr_return = (nvidia_prices[day+1] - nvidia_prices[day]) / nvidia_prices[day]
    nvidia_returns.append(curr_return)

amd_volatility = statistics.stdev(amd_returns)
nvidia_volatility = statistics.stdev(nvidia_returns)

print(f"AMD volatility: {amd_volatility:.4f}")
print(f"NVIDIA volatility: {nvidia_volatility:.4f}")
```

### Preliminary Observations

- Returns allow us to compare price movements relative to each asset's starting price.
- Standard deviation measures the dispersion of returns and serves as a measure of volatility.
- Volatility alone does not determine whether an asset is a better investment.
- These hypothetical observations are illustrative and do not represent actual market data.
